```python
import numpy as np
import pandas as pd
```

FInding Missing Data

```python
data = {
          'A': [1, 2, np.nan, 4, 5],
          'B': [1, 2, 3, 4, 5],
          'C': [1, 2, 3, np.nan, np.nan],
          'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna()  #Detects missing values across the entire DataFrame.
```

```python
df.isna().sum() #Counts the total number of missing values per column.
```

```python
df.isna().any() #Checks whether any missing values exist per column.
```

Removing Missing Data

```python
df.dropna() #By default for rows
```

```python
df.dropna(axis = 0) #For rows
```

```python
df.dropna(axis = 1) #For columns
```

```python
df
```

```python
df.dropna(thresh = 3) #The thresh parameter specifies the minimum number of non-NaN values required for a row (or column) to be kept.
```

Filling The Missing Data

```python
df
```

```python
df.fillna(0)
```

```python
values = {'A':0, 'B':100, 'C':200, 'D':300}
df.fillna(value = values)
```

```python
df.fillna(df.mean())
```

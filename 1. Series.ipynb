## Pandas Series
A series is a one-dimensional labelled arraycapable of holding any data type. The axis labels are collectively called the index

## Creating Series
There are multiple ways to create pandas series

```python
import numpy as np
import pandas as pd
```

```python
labels = ['a','b','c']
my_list = [10,20,30]
my_arr = np.array([30,40,50])
my_dict = {1:80, 2:90, 3:100}
```

```python
pd.Series(my_list, index = labels)
```

```python
pd.Series(my_dict)
```

```python
pd.Series(my_arr, index = my_list)
```

```python
pd.Series(my_arr, index = labels)
```

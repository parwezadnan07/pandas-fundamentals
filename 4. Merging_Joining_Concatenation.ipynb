Merging, Joining & Concatenation

```python
import numpy as np
import pandas as pd
```

Merginig 2 DataFrames

```python
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2 : Salary Infromation
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
employees
```

```python
salaries
```

```python
# Inner Join
```

```python
df1 = pd.merge(employees, salaries, on = 'employee_id', how = 'inner') 
df1
```

```python
# Full or Outer Join
```

```python
df2 = pd.merge(employees, salaries, on = 'employee_id', how = 'outer')
df2
```

```python
# Left Join
```

```python
df3 = pd.merge(employees, salaries, on = 'employee_id', how = 'left')
df3
```

```python
# Right Join
```

```python
df4 = pd.merge(employees, salaries, on = 'employee_id', how = 'right')
df4
```

Concatenation of 2 DataFrames

```python
df5 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df6 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
df5
```

```python
df6
```

```python
pd.concat([df5, df6])
```

```python
pd.concat([df5, df6], axis = 1)
```

Joining 2 DataFrames

```python
df7 = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie']
}, index=[1, 2, 3])

# Second DataFrame
df8 = pd.DataFrame({
    'score': [85, 90, 75]
}, index = [2, 3, 4])
```

```python
df7
```

```python
df8
```

```python
df7.join(df8)
```

```python
df7.join(df8, how = 'outer')
```
